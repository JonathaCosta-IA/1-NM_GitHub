### Calc of numerical precision


```python
import numpy as np
a=1.
b=-100.0001
c=0.01

delta = b**2 - 4*a*c
raiz_delta = delta**0.5

x1 = (-b + raiz_delta) / (2*a)
x2 = (-b - raiz_delta) / (2*a)  
x1,x2
```

```python
# Novo exercício:
x=np.array([10,1e3,1e5])
f = lambda x: x*(x**0.5 - (x-1)**(0.5))
for i in x:
    print(f"x={i:.0f} -> f(x)={f(i):.16f}")
```
