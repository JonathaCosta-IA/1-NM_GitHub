## Conceitos Básicos

### Conversão de Binário para Decimal

A conversão de um número binário (base 2) para decimal (base 10) consiste em representar seu valor numérico por meio da soma ponderada de seus dígitos, utilizando potências de 2.

Considere o número binário: $num_2 = (101)_2$

O processo de conversão pode ser representado por:
$num_{10} = \operatorname{converter}(num_2) \Rightarrow num_{10} = \operatorname{converter}(101_2)$

### Método Convencional

Cada dígito binário possui um peso correspondente a uma potência de 2, determinado por sua posição no número.

Para o exemplo $101_2$, temos:

$
num_{10} = 1\cdot 2^2 + 0\cdot 2^1 + 1\cdot 2^0 \Rightarrow num_{10} = 4 + 0 + 1 = 5 
\Rightarrow \boxed{(101)_2 = (5)_{10}}
$

### Formalização Matemática

Considere um número binário representado por uma sequência de $n$ dígitos:

$
num_2 = b_0b_1b_2\ldots b_{n-1} \Rightarrow
\boxed{
num_{10} = \sum_{i=0}^{n-1} b_i \cdot 2^{n-1-i}
}
$

Em que:
- $num_{10}$: valor numérico na base decimal;
- $b_i$: dígito binário na posição $i$, sendo $b_i \in \{0,1\}$;
- $n$: quantidade total de dígitos do número binário;
- $i$: índice da posição do dígito, iniciando em zero.

**Observação:** Note que o expoente da potência de 2 é determinado pela posição do dígito, considerando que o último dígito possui expoente zero.


## Conversão de binário para decimal

### Conversão de binário sem mantissa para decimal

```python
# Conversão de binário para decimal
#****************************************
num_2 = "101101"

# Método convencional utilizando a equação 
'''
num_{10} = sum_{i=0}^{n-1} b_i * 2^{n-1-i}
'''
n = len(num_2)
soma=0
for i in range(n):       # Note que o comando 'range' encerra o incremento em (n-1).Extremo superior no somatário da equação.
    soma += int(num_2[i]) * 2**(n-1-i)
print(f"O número binário {num_2} vale {soma} em decimal.")

#****************************************
# Convertendo essa estrutura numa função para reutilizá-la, temos:
def bin2dec(num:str):
    n=len(num)
    soma=0
    for i in range(n):
        soma += int(num[i]) * 2**(n-1-i)

    return soma
#****************************************
# Logo:
print(f"O número binário {num_2} vale {bin2dec(num_2)} em decimal.")

# Nota:
# A funçao nativa no python para essa operação é int(num_2,2)
```

### Conversão de binário com mantissa para decimal

```python
# Procedimento: separar o número para converter cada parte, juntando as partes convertidas ao término

n2 = "11010,101"

def bin2dec_frac(num:str):
    # Separando a parte inteira e a fracionária
    inteira_bin, fracionaria_bin = num.split(",") 
    
    # Convertendo a parte fracionária (mantissa)
    fracionaria_dec = 0.0
    n=len(fracionaria_bin)
    for i in range(n):
        fracionaria_dec += int(fracionaria_bin[i]) * 2 **(-(i+1))   # Para (0,101)  -> 1*2^-1 + 0*2^-2 + 1*2^-3
    
    # Convertendo a parte inteira
    inteira_dec = bin2dec(inteira_bin)
    # Numero convertido
    num = inteira_dec + fracionaria_dec
    
    return num

print(bin2dec_frac(n2))
```

## Conversão de decimal para binário

### Conversão de decimal inteiro para binário¶

```python
# Método convencional utilizando sucessivas divisões 
n10 = 16

n2 =""                
D=n10
while(True):            # Ciclo de sucessivas divisões
    if D==1:
        n2+=str(1)
        break
    n2+=str(D%2)
    D=D//2              # Atualiza o dividendo
n2=n2[::-1]             # Ordenando o bits

#*****************************************
#Gerando uma função a partir desse rotina
def dec2bin(num:int):
    n2 =""                
    D=num
    while(True):          #  Ciclo de sucessivas divisões
        if D==1:
            n2+=str(1)
            break
        n2+=str(D%2)
        D=D//2            # Atualiza o dividendo
    n2=n2[::-1]           # Ordenando o bits
    return n2
```

```python
for num in [45,2345]:
        print(f"O número decimal {num} vale {dec2bin(num)} em binário.")
```

## Representação numérica

```python
'''
Para x = 0, 1, obtenha a representação binária aproximada utilizando 4 bits (0, 0001)2 , 8 bits (0, 00011001)2
e 16 bits (0, 0001100110011001)2.
Para cada aproximação x̂, calcule:
Ea = |x − x̂|
'''

# Alterando a função DecFrac2bin para retornar com 'n' bits
def DecFrac2bin_Bits(NumDecFrac:float,n:int):
    bits=1
    NumBin ="0,"               
    while(True):                            # Ciclo de sucessivas multiplicações
        NumDecFrac *= 2                     # Multiplica por 2
        NumBin+= str(int(NumDecFrac))       # Guarda a parte inteira
        if NumDecFrac >= 1 : NumDecFrac-=1  # Remove o digito a esquerda caso seja maior que 1 
        if NumDecFrac == 0: break
        if bits==n: 
            break
        else:
            bits+=1    
        
    return NumBin
    
```

```python
num=0.1

for bits in [4, 8, 16,32]:
    x1 = DecFrac2bin_Bits(num, bits)  # Representação binária aproximada
    x2 = bin2dec_frac(x1)             # Reconversão para decimal

    Ea = abs(num - x2)                # Erro absoluto

    print(f"Bits = {bits}")
    print(f"Binário = {x1}")
    print(f"Decimal = {x2:.10f}")
    print(f"Erro absoluto = {Ea:.10f}")
    print("-" * 40)
```

```python
# Questão alternativa
x = 0.1
for n in [4, 8, 16, 32, 64]:
    valor = x
    aproximacao = 0
    for i in range(1, n + 1):
        valor *= 2
        bit = int(valor)
        aproximacao += bit / 2**i
        valor -= bit
    erro = abs(x - aproximacao)
    print(f"{n} bits")
    print(f"aproximacao = {aproximacao}")
    print(f"erro = {erro}\n")

print("Analise os resultados e expresse suas conclusões!")
```
