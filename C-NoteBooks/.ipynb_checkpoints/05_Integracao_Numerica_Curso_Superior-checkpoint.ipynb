# Laboratório computacional — Integração Numérica

A seguir, as principais regras apresentadas no material são transformadas em **funções executáveis**, mantendo a distinção entre métodos simples e compostos. Esta seção é complementar ao conteúdo original.

```python
def retangulo_esquerdo(f, a, b):
    return f(a) * (b - a)

def ponto_central(f, a, b):
    return f((a + b) / 2) * (b - a)

def trapezio_simples(f, a, b):
    return (b - a) * (f(a) + f(b)) / 2

def simpson_13(f, a, b):
    h = (b - a) / 2
    return h / 3 * (f(a) + 4*f(a+h) + f(b))

def simpson_38(f, a, b):
    h = (b - a) / 3
    return 3*h/8 * (f(a) + 3*f(a+h) + 3*f(a+2*h) + f(b))
```

### Exemplo didático complementar

Para validar as implementações, considere $f(x)=x^2$ no intervalo $[0,2]$. A integral exata é

$$
I=\int_0^2 x^2\,dx=rac{8}{3}.
$$

```python
f = lambda x: x**2
a, b = 0.0, 2.0
exata = 8/3

resultados = {
    "Retângulo esquerdo": retangulo_esquerdo(f, a, b),
    "Ponto central": ponto_central(f, a, b),
    "Trapézio simples": trapezio_simples(f, a, b),
    "Simpson 1/3": simpson_13(f, a, b),
    "Simpson 3/8": simpson_38(f, a, b),
}

for metodo, valor in resultados.items():
    erro = abs(valor - exata) / abs(exata) * 100
    print(f"{metodo:22s}: {valor:.10f} | erro = {erro:.6f}%")
```

```python
def trapezio_composto(f, a, b, n):
    x = np.linspace(a, b, n + 1)
    h = (b - a) / n
    return h * (0.5*f(x[0]) + np.sum(f(x[1:-1])) + 0.5*f(x[-1]))

# Estudo da convergência
f = lambda x: np.exp(-x**2)
a, b = 0.0, 1.0
ns = np.array([4, 8, 16, 32, 64, 128])
valores = np.array([trapezio_composto(f, a, b, int(n)) for n in ns])
print(np.c_[ns, valores])
```

## Integração Numérica Introdução-conceitos

Introdução

- Os métodos de integração numérica aproximam valores de integrais definidas.
- A integração numérica é útil quando:
      - Não se conhece a função f (x), mas tem-se apenas uma tabela de valores para f ;
      - A função f é conhecida, mas é muito complexa, o que dificulta a determinação da primitiva.
- A ideia básica é substituir f (x) por um polinômio que aproxime-a razoavelmente no
     intervalo [a, b];
- Assim o problema se resume à integração de polinômios, o que é trivial de realizar.


## Integração Numérica Introdução-conceitos

Introdução
Existem vários métodos disponíveis para o cálculo numérico de integrais. Em cada um desses métodos, uma
fórmula é deduzida para calcular o valor aproximado de uma integral a partir dos pontos fechado discretos
do integrando.

  - Métodos fechados - os pontos finais do intervalo (e o
      integrando) são utilizados na fórmula que estima o valor
      da integral.
      - Método trapezoidal
      - Método de Simpson
  - Métodos abertos - o intervalo de integração se estende
      além do limite especificado pelos pontos finais
      - Método do ponto central

## Integração Numérica Métodos de Simpson

Métodos de Simpson

  - O método trapezoidal aproxima o integrando por uma linha reta. Em tese, uma melhor aproximação
      pode ser obtida com a representação do integrando como uma função não-linear de fácil integração.

  - Uma classe de métodos com essa característica (regras ou métodos de Simpson) utiliza para aproximar
      o integrando polinômios:
      - quadráticos (método de Simpson 1/3)
      - cúbicos (método de Simpson 3/8)


## Integração Numérica Métodos de Simpson

Método de 1/3 de Simpson Simples
  - Um polinômio quadrático (de segunda ordem) é utilizado
       para aproximar o integrando
  - Os coeficientes do polinômio quadrático são determinados
       a partir de três pontos: x1 = a, x2 = (a + b)/2 e x3 = b
       tal que p(x) = α + β(x - x1 ) + γ(x - x1 )(x - x2 ) em que
       α,β e γ são constantes desconhecidas tais que o polinômio
       deve passar por todos os pontos, p(x1 ) = f (x1 ), p(x2 ) =
       f (x2 ) e p(x3 ) = f (x3 ).
  - O que resultado em:
       α = f (x1 ),β = [f (x2 ) - f (x1 )]/(x2 - x1 ) e γ =
       f (x3 ) - 2f (x2 ) + f (x1 )

## Integração Numérica Métodos de Simpson

Método de 1/3 de Simpson Composto

  - No método de Simpson 1/3 composto divide-se o intervalo [a, b] em N subintervalos de largura h, em
      que h = (b - a)/N
  - O método de Simpson 1/3 é aplicado em dois subintervalos adjacentes de cada vez, por serem necessários
      3 pontos para definir um polinômio quadrático. Portanto, o intervalo total deve ser dividido em um
      número par de subintervalos.
  - A integral ao longo de 2 intervalos adjacentes [xi-1 , xi ] e [xi , xi+1 ] fica definida como: Ii (f ) =
      Z xi+1
                      h
             f (x)dx = [f (xi-1 ) + 4f (xi ) + f (xi+1 )], onde h = xi+1 - xi = xi - xi-1
       xi-1           3

  - Condições necessárias
      - Os subintervalos devem ser igualmente espaçados.
      - O número de subintervalos no domínio [a, b] deve ser um número par

Método de 3/8 de Simpson Composto
  - No método de Simpson 3/8 composto divide-se o intervalo [a, b] em N subintervalos de largura h, em
      que h = (b - a)/N
  - Sendo necessários 4 pontos para definir um polinômio cúbico, o método é aplicado em três subin-
      tervalos adjacentes de cada vez e, portanto, o número de subintervalos dever ser divisível por 3.
  - A integral ao longo de 3 intervalos adjacentes para 4 grupos fica definida como:
