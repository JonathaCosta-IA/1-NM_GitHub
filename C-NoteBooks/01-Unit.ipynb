01. Calcule a função:
$
\displaystyle
f(x) = \lim_{x \to 0} \frac{\sin(x)}{x}
$

```python
import sympy as sp
x = sp.Symbol('x')
f=sp.sin(x)/x
sp.limit(f,x,0) # Limite de f(x) para x -> 0.
```

02. Calcule $f'(x)$ para
$
\displaystyle
f(x) = x^3\cdot sin(x)
$

```python
import sympy as sp
x = sp.Symbol('x')
g = x**3 * sp.sin(x)
sp.diff(g,x)
```

### Limites

```python
# 01. Limite sin(x) / x, com x->0
f= sp.sin(x)/ x
limit_f = sp.limit(f,x,0)
print(f"O limite de f com x -> 0 é {limit_f}.")

# 02. Limite (x**2 - 4)/(x-1),com  x->2
f1 = (x**2 - 4)/(x-1)
lim_f1 = sp.limit(f1,x,2)
print(f"O limite de {f1} com x -> 2 é {lim_f1}.")

# 03. Limite (x**3 - 1)/(x-1),com  x->1
f2 = (x**3 - 1)/(x-1)
lim_f2 = sp.limit(f2,x,1)
print(f"O limite de {f2} com x -> 1 é {lim_f2}.")

# 04. Limite 1-cos(x) / x**2,com  x->0
f3 = 1-sp.cos(x) / x**2
lim_f3 = sp.limit(f3,x,0)
print(f"O limite de {f3} com x -> 0 é {lim_f3}.")
```

### Teoremas do valor intermediário

```python
x = sp.Symbol('x')                      # Variável simbólica
f = x**2 - 2                            # Função usando 'x' simbolico
a, b = 1, 2                             # Intervalo em análise

fa, fb = f.subs(x,a), f.subs(x,b)       # Avalia nos extremos
print("f(a) =", fa, "| f(b) =", fb)

print("Existe raiz:", fa*fb < 0)        # Verifica mudança de sinal. Condição sine qua no.
print("Raiz:", sp.solve(f, x))          # Calcula as raízes
```

```python
x, c = sp.symbols('x c')
f = x**2
df = sp.diff(f, x)                      # Calcula a derivada
a, b = 1, 3
tm = (f.subs(x,b)-f.subs(x,a))/(b-a)   # (Delta Y) / (Delta X): taxa média de variação

c = sp.solve(df.subs(x,c)-tm, c)       # Determina o ponto c
print("Taxa média:", tm)
print("Ponto c:", c)
```

```python
f = x**3 - 3*x**2 + 2*x + 1
df = sp.diff(f, x)                      # Primeira derivada
pontos = sp.solve(df, x)                # Pontos críticos: raízes

print("f'(x) =", df)
print("Pontos críticos:", pontos)

for p in pontos:
    print(f"\nf(x) no ponto crítico:\n {p} vale {f.subs(x,p)}")
```

```python
f = x**2 + 1
I = sp.integrate(f, (x, 0, 2))         # Calcula a integral definida

print("Integral:", I)
print("Valor numérico:", float(I))
```

```python
f = 3*x**2 + 2*x + 1
F = sp.integrate(f, x)                  # Calcula a antiderivada

I = sp.integrate(f, (x, 0, 2))         # Integral definida
dF = F.subs(x,2) - F.subs(x,0)         # Diferença nos extremos

print("Antiderivada:", F)
print("Integral definida:", I)
print("F(b) - F(a):", dF)
print("Verificação:", sp.simplify(I-dF) == 0)
```
